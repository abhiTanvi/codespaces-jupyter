```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt
import seaborn as sns


url = 'https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud'
  # Replace with actual file URL
df = pd.read_csv(url)

X = df.drop(['Time', 'Class'], axis=1)  # Exclude 'Time' and 'Class' from features
y = df['Class']

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

# 2. Temporal Feature Engineering
df['HourOfDay'] = df['Time'].apply(lambda x: (x // 3600) % 24) 

# 3. Feature Scaling (Standardization)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_resampled)


df_sorted = df.sort_values('Time')
train_size = int(0.8 * len(df_sorted))

train_data = df_sorted[:train_size]
test_data = df_sorted[train_size:]

X_train = train_data.drop(['Class'], axis=1)
y_train = train_data['Class']
X_test = test_data.drop(['Class'], axis=1)
y_test = test_data['Class']

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Training size: {len(X_train)}, Test size: {len(X_test)}")
```
